```python
### TextLoader->> This is used to load the text file and create a document object. It reads the content of the text file and creates a Document object with the content and metadata.
### Basically it lets you to read the text file

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/sample.txt",encoding="utf-8")
loader.load() # to get the components of the document object, we can use the load() method. It returns a list of Document objects. 
```

```python
### Directory Loader --> This is used for loading all the files present in the Text_files folder.

from langchain_community.document_loaders import DirectoryLoader

## Load all the text files in the directory
dir_loader=DirectoryLoader(
    "../data/text_files", 
    glob="*.txt", 
    loader_cls=TextLoader, 
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)

documents=dir_loader.load() # to get the components of the document object, we can use the load() method. It returns a list of Document objects.
documents
```
